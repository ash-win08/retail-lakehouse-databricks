```python
import os, glob, shutil

dbutils.widgets.text("month", "2018-03")
month = dbutils.widgets.get("month")

SRC = "/Volumes/workspace/bronze/landing/daily"      # full archive (stays untouched)
DST = "/Volumes/workspace/bronze/landing/incoming"   # what the pipeline reads

n = 0
for t in ["orders", "order_items", "order_payments", "customers"]:
    os.makedirs(f"{DST}/{t}", exist_ok=True)
    for f in glob.glob(f"{SRC}/{t}/{t}_{month}-*.csv"):
        shutil.copy(f, f"{DST}/{t}/{os.path.basename(f)}")
        n += 1
print(month, n, "files delivered")
```
